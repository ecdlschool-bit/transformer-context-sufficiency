# Paper Results Reproduction

Unified analysis notebook for regenerating the reported tables, figures, and summary statistics from archived experiment outputs.

This workflow performs analysis reproduction only: it does not retrain models, repeat pretrained-model extraction, rerun interventions, or regenerate the HMM simulations.

A successful run validates the archived inputs, regenerates the paper-level outputs, and executes the numerical integrity checks.


````python

# ============================================================
# 0. Imports, folder-aware file resolver, output directories
# ============================================================
from pathlib import Path
import io, json, re, zipfile, shutil, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

warnings.filterwarnings("ignore", category=RuntimeWarning)

ROOT = Path.cwd()
OUTPUT = ROOT / "Reproduced_Paper_Outputs"
TABLE_DIR = OUTPUT / "tables"
FIG_DIR = OUTPUT / "figures"
AUDIT_DIR = OUTPUT / "audit"
for d in [OUTPUT, TABLE_DIR, FIG_DIR, AUDIT_DIR]:
    d.mkdir(parents=True, exist_ok=True)

def normalized_name(name):
    p = Path(name)
    stem = re.sub(r"\s+\(\d+\)$", "", p.stem)
    return stem.lower() + p.suffix.lower()

def resolve_file(canonical_name, required=True):
    """
    Resolve a result archive from the notebook folder OR the existing
    Markov_Context_Paper project tree.

    This keeps the user's current directory structure intact; files do not
    need to be copied next to the notebook.
    """
    target = normalized_name(canonical_name)

    search_dirs = [
        ROOT,
        ROOT / "results",
        ROOT / "Markov_Context_Paper",
        ROOT / "Markov_Context_Paper" / "Part1",
        ROOT / "Markov_Context_Paper" / "Part2",
        ROOT / "Markov_Context_Paper" / "Reviewer_Robustness_Checks",
        ROOT / "Markov_Context_Paper" / "Reviewer_A3b_TrueState_Coverage",
        ROOT / "Markov_Context_Paper" / "HMM_Latent_State_Extension",
        ROOT / "Markov_Context_Paper" / "HMM_Oracle_Gap_Fix_GPU",
    ]

    # Also support the common case where the notebook itself is already
    # inside Markov_Context_Paper or one of its subdirectories.
    search_dirs += [
        ROOT / "Part1",
        ROOT / "Part2",
        ROOT / "Reviewer_Robustness_Checks",
        ROOT / "Reviewer_A3b_TrueState_Coverage",
        ROOT / "HMM_Latent_State_Extension",
        ROOT / "HMM_Oracle_Gap_Fix_GPU",
    ]

    # Deduplicate while preserving order.
    seen = set()
    search_dirs = [d for d in search_dirs if not (str(d.resolve()) in seen or seen.add(str(d.resolve())))]

    matches = []
    for d in search_dirs:
        if not d.exists() or not d.is_dir():
            continue
        # Exact filename first.
        exact = d / canonical_name
        if exact.exists():
            return exact
        # Then accept duplicate-download suffixes such as (2), (3), ...
        for p in d.iterdir():
            if p.is_file() and normalized_name(p.name) == target:
                matches.append(p)

    if matches:
        return sorted(matches, key=lambda p: (len(p.parts), p.name))[-1]

    if required:
        searched = "\n  - ".join(str(d) for d in search_dirs)
        raise FileNotFoundError(
            f"Missing required file: {canonical_name}\nSearched:\n  - {searched}"
        )
    return None

def zip_read_csv(zpath, member):
    with zipfile.ZipFile(zpath, "r") as z:
        return pd.read_csv(io.BytesIO(z.read(member)))

def zip_read_json(zpath, member):
    with zipfile.ZipFile(zpath, "r") as z:
        return json.loads(z.read(member).decode("utf-8"))

def zip_find_member(zpath, basename):
    with zipfile.ZipFile(zpath, "r") as z:
        hits = [n for n in z.namelist() if Path(n).name == basename]
    if not hits:
        raise FileNotFoundError(f"{basename} not found inside {zpath.name}")
    if len(hits) > 1:
        # prefer shallowest path
        hits = sorted(hits, key=lambda x: (x.count("/"), x))
    return hits[0]

print("Working folder:", ROOT)
print("Output folder :", OUTPUT)
````

````python

# ============================================================
# 3A. Final Part I / Part II publication tables
# ============================================================
MF = FILES["markov_final"]

main_summary = zip_read_csv(MF, "csv/part1_main_threshold_summary.csv")
threshold_contrasts = zip_read_csv(MF, "csv/part1_threshold_contrasts.csv")
k4_refinement = zip_read_csv(MF, "csv/part1_k4_refinement_summary.csv")
capacity_summary = zip_read_csv(MF, "csv/part1_capacity_summary.csv")
sparse_summary = zip_read_csv(MF, "csv/part1_sparse_summary.csv")
vocab_summary = zip_read_csv(MF, "csv/part1_vocab_summary.csv")
v32_targeted = zip_read_csv(MF, "csv/part1_v32k3_targeted_summary.csv")

p2_final_summary = zip_read_csv(MF, "csv/part2_final_linear_summary.csv")
p2_threshold_contrasts = zip_read_csv(MF, "csv/part2_threshold_contrasts.csv")
p2_model_size = zip_read_csv(MF, "csv/part2_model_size_comparison.csv")
p2_linear_vs_mlp = zip_read_csv(MF, "csv/part2_linear_vs_mlp.csv")
control_summary = zip_read_csv(MF, "csv/part2_input_vs_state_control_summary_final.csv")
cross_part = zip_read_csv(MF, "csv/cross_part_synthesis.csv")

# Raw Part II is retained for the layerwise figure.
P2 = FILES["part2"]
p2 = zip_read_csv(P2, "exports/csv/part2_probe_results_raw.csv")


# Recompute corrected publication summaries from archived per-seed data.
baseline = pd.read_csv(resolve_file("Part2_THEORY_BASELINE_EXACT_AUDIT.csv"))
assert len(baseline) == 90 and not baseline.duplicated(["k","L","seed"]).any()
p2 = p2.merge(baseline[["k","L","seed","corrected_Hstar_stationary",
    "corrected_HL_stationary","corrected_info_gap"]], on=["k","L","seed"],
    how="left", validate="many_to_one")
assert not p2["corrected_HL_stationary"].isna().any()
p2["H_star"] = p2["corrected_Hstar_stationary"]
p2["H_L"] = p2["corrected_HL_stationary"]
p2["info_gap"] = p2["corrected_info_gap"]
p2["next_excess_over_HL"] = p2["next_ce"] - p2["H_L"]
part2_raw = p2.copy()
CSV_DIR = TABLE_DIR

# ============================================================
# 3. Bootstrap CI and effect-size helpers
# ============================================================

def bootstrap_mean_ci(values, n_boot=5000, ci=0.95, seed=123):
    x = np.asarray(pd.Series(values).dropna(), dtype=float)
    if len(x) == 0:
        return (np.nan, np.nan, np.nan)
    if len(x) == 1:
        return (float(x[0]), np.nan, np.nan)

    rng = np.random.default_rng(seed)
    boots = rng.choice(x, size=(n_boot, len(x)), replace=True).mean(axis=1)
    alpha = (1-ci)/2
    return (
        float(x.mean()),
        float(np.quantile(boots, alpha)),
        float(np.quantile(boots, 1-alpha)),
    )

def paired_bootstrap_diff(a, b, n_boot=5000, ci=0.95, seed=123):
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)
    mask = np.isfinite(a) & np.isfinite(b)
    d = a[mask] - b[mask]
    return bootstrap_mean_ci(d, n_boot=n_boot, ci=ci, seed=seed)

def cohens_d_paired(a, b):
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)
    mask = np.isfinite(a) & np.isfinite(b)
    d = a[mask] - b[mask]
    if len(d) < 2 or np.std(d, ddof=1) == 0:
        return np.nan
    return float(np.mean(d) / np.std(d, ddof=1))

def summarize_group(df, group_cols, value_col):
    rows = []
    for keys, g in df.groupby(group_cols):
        if not isinstance(keys, tuple): keys = (keys,)
        mean, lo, hi = bootstrap_mean_ci(g[value_col].values)
        row = dict(zip(group_cols, keys))
        row.update({
            "mean": mean,
            "ci95_low": lo,
            "ci95_high": hi,
            "sd": float(g[value_col].std(ddof=1)) if len(g) > 1 else np.nan,
            "n": len(g),
        })
        rows.append(row)
    return pd.DataFrame(rows)


# ============================================================
# 9. Part 2 primary summaries
# ============================================================

p2 = part2_raw.copy()
p2["relative_context"] = p2["L"] - p2["k"]

# final linear layer per model
final_linear = []
for mk, g in p2[p2.probe_type=="linear"].groupby("model_key"):
    max_layer = g.layer_index.max()
    final_linear.append(g[g.layer_index == max_layer])
final_linear = pd.concat(final_linear, ignore_index=True)

p2_final_summary = (
    final_linear.groupby(["model_key","k","L","relative_context"], as_index=False)
                .agg(
                    state_exact_mean=("state_exact_acc","mean"),
                    state_exact_sd=("state_exact_acc","std"),
                    oldest_acc_mean=("state_oldest_acc","mean"),
                    next_ce_mean=("next_ce","mean"),
                    next_excess_HL_mean=("next_excess_over_HL","mean"),
                    next_excess_HL_sd=("next_excess_over_HL","std"),
                    info_gap_mean=("info_gap","mean"),
                    n=("seed","count")
                )
)

p2_final_summary.to_csv(CSV_DIR / "part2_final_linear_summary.csv", index=False)

# ============================================================
# 10. Part 2 L=k-1 -> L=k paired contrasts
# ============================================================

rows = []

for mk in sorted(final_linear.model_key.unique()):
    for k in sorted(final_linear.k.unique()):
        b = final_linear[
            (final_linear.model_key==mk)&
            (final_linear.k==k)&
            (final_linear.L==k-1)
        ][["seed","state_exact_acc","state_oldest_acc","next_excess_over_HL"]]

        a = final_linear[
            (final_linear.model_key==mk)&
            (final_linear.k==k)&
            (final_linear.L==k)
        ][["seed","state_exact_acc","state_oldest_acc","next_excess_over_HL"]]

        m = b.merge(a,on="seed",suffixes=("_below","_at"))

        # for recovery, report at-minus-below
        for metric in ["state_exact_acc","state_oldest_acc"]:
            mean,lo,hi = paired_bootstrap_diff(
                m[f"{metric}_at"], m[f"{metric}_below"]
            )
            rows.append({
                "model_key":mk,"k":k,"metric":metric,
                "contrast":"at_minus_below",
                "mean_diff":mean,"ci95_low":lo,"ci95_high":hi,
                "paired_cohens_d":cohens_d_paired(
                    m[f"{metric}_at"],m[f"{metric}_below"]
                ),
                "n":len(m)
            })

        # for next excess, lower is better; report below-minus-at
        metric="next_excess_over_HL"
        mean,lo,hi = paired_bootstrap_diff(
            m[f"{metric}_below"],m[f"{metric}_at"]
        )
        rows.append({
            "model_key":mk,"k":k,"metric":metric,
            "contrast":"below_minus_at",
            "mean_diff":mean,"ci95_low":lo,"ci95_high":hi,
            "paired_cohens_d":cohens_d_paired(
                m[f"{metric}_below"],m[f"{metric}_at"]
            ),
            "n":len(m)
        })

p2_threshold = pd.DataFrame(rows)
p2_threshold.to_csv(CSV_DIR / "part2_threshold_contrasts.csv", index=False)

# ============================================================
# 12. Final-layer linear vs MLP
# ============================================================

lin = p2[p2.probe_type=="linear"].copy()
mlp = p2[p2.probe_type=="mlp_final"].copy()

# retain only final linear layer per model
lin_final=[]
for mk,g in lin.groupby("model_key"):
    lin_final.append(g[g.layer_index==g.layer_index.max()])
lin_final=pd.concat(lin_final,ignore_index=True)

lm = lin_final.merge(
    mlp,
    on=["model_key","model_id","k","L","seed"],
    suffixes=("_linear","_mlp")
)

mlp_rows=[]
for (mk,k,L),g in lm.groupby(["model_key","k","L"]):
    for metric in ["state_exact_acc","state_oldest_acc"]:
        mean,lo,hi=paired_bootstrap_diff(
            g[f"{metric}_mlp"],g[f"{metric}_linear"]
        )
        mlp_rows.append({
            "model_key":mk,"k":k,"L":L,"metric":metric,
            "MLP_minus_linear":mean,"ci95_low":lo,"ci95_high":hi,"n":len(g)
        })

    mean,lo,hi=paired_bootstrap_diff(
        g["next_excess_over_HL_linear"],
        g["next_excess_over_HL_mlp"]
    )
    mlp_rows.append({
        "model_key":mk,"k":k,"L":L,"metric":"next_excess_over_HL",
        "MLP_minus_linear":mean, # positive here = linear worse than MLP
        "ci95_low":lo,"ci95_high":hi,"n":len(g)
    })

mlp_cmp=pd.DataFrame(mlp_rows)
mlp_cmp.to_csv(CSV_DIR/"part2_linear_vs_mlp.csv",index=False)

p2_threshold_contrasts = p2_threshold
p2_linear_vs_mlp = mlp_cmp
# Checkpoint/token maps were sampled independently across model sizes:
# report descriptive differences only, without a token-paired confidence interval.
model_rows = []
for (k,L), g in final_linear.groupby(["k","L"]):
    for metric in ["state_exact_acc", "state_oldest_acc", "next_excess_over_HL"]:
        means = g.groupby("model_key")[metric].mean()
        model_rows.append(dict(k=k,L=L,metric=metric,
            comparison="1.5B_minus_0.5B", mean_diff=means["qwen15"]-means["qwen05"],
            interpretation="descriptive; independently sampled token maps"))
p2_model_size = pd.DataFrame(model_rows)
shuffle_zip = resolve_file("Part2_ShuffledLabel_Corrected_10seed_RESULTS.zip")
shuffle = zip_read_csv(shuffle_zip, zip_find_member(shuffle_zip,
    "corrected_shuffled_label_summary_10seed.csv"))
control_summary = control_summary.drop(columns=["shuffle_exact_mean","shuffle_pos_mean"]).merge(
    shuffle[["model_key","k","L","shuffle_exact_mean","shuffle_pos_mean"]],
    on=["model_key","k","L"], validate="one_to_one")
for idx, row in cross_part.iterrows():
    for suffix,L in [("below",row.k-1),("at",row.k)]:
        selected = final_linear[(final_linear.model_key==row.model_key)&
            (final_linear.k==row.k)&(final_linear.L==L)]
        cross_part.loc[idx,"part2_next_excess_HL_"+suffix] = selected.next_excess_over_HL.mean()
mc_raw = pd.read_csv(resolve_file("reviewer_R1_5_HL_uncertainty_raw.csv"))
mc_summary = mc_raw.groupby(["V","k","L","source_seed","N"],as_index=False).agg(
    gap_mean=("info_gap","mean"),mc_sd=("info_gap","std"),n=("rep","count"))
mc_summary["mc_se"] = mc_summary.mc_sd / np.sqrt(mc_summary.n)
mc_summary["normal95_halfwidth"] = 1.96 * mc_summary.mc_se
mc_summary.to_csv(TABLE_DIR/"monte_carlo_within_kernel_uncertainty.csv",index=False)
mc_primary = mc_summary[mc_summary.N==500000]
p2.to_csv(TABLE_DIR/"part2_probe_results_stationary_corrected.csv",index=False)


p1_pub = zip_read_csv(MF, "csv/part1_publication_results_clean.csv")

# ============================================================
# 4. Main threshold summary
# ============================================================

main = p1_pub[p1_pub.experiment == "main_recovery"].copy()

main_summary = (
    main.groupby(["k","L"], as_index=False)
        .agg(
            test_loss_mean=("test_loss","mean"),
            test_loss_sd=("test_loss","std"),
            H_star_mean=("H_star","mean"),
            H_L_mean=("H_L","mean"),
            info_gap_mean=("info_gap","mean"),
            info_gap_sd=("info_gap","std"),
            model_gap_mean=("model_gap","mean"),
            model_gap_sd=("model_gap","std"),
            total_excess_mean=("total_excess","mean"),
            total_excess_sd=("total_excess","std"),
            n=("seed","count"),
        )
)

# Add bootstrap CI for model gap
mg_ci = summarize_group(main, ["k","L"], "model_gap")
main_summary = main_summary.merge(
    mg_ci[["k","L","ci95_low","ci95_high"]],
    on=["k","L"], how="left"
).rename(columns={
    "ci95_low":"model_gap_ci95_low",
    "ci95_high":"model_gap_ci95_high"
})

main_summary.to_csv(CSV_DIR / "part1_main_threshold_summary.csv", index=False)

# ============================================================
# 5. Paired threshold contrasts by seed
# ============================================================

threshold_rows = []

for k in sorted(main.k.unique()):
    below = main[(main.k == k) & (main.L == k-1)][["seed","total_excess","info_gap","model_gap"]]
    at = main[(main.k == k) & (main.L == k)][["seed","total_excess","info_gap","model_gap"]]

    m = below.merge(at, on="seed", suffixes=("_below","_at"))

    for metric in ["total_excess","info_gap","model_gap"]:
        mean, lo, hi = paired_bootstrap_diff(
            m[f"{metric}_below"].values,
            m[f"{metric}_at"].values
        )
        d = cohens_d_paired(
            m[f"{metric}_below"].values,
            m[f"{metric}_at"].values
        )
        threshold_rows.append({
            "k": k,
            "metric": metric,
            "below_L": k-1,
            "at_L": k,
            "mean_drop_below_minus_at": mean,
            "ci95_low": lo,
            "ci95_high": hi,
            "paired_cohens_d": d,
            "n": len(m),
        })

threshold_contrasts = pd.DataFrame(threshold_rows)
threshold_contrasts.to_csv(CSV_DIR / "part1_threshold_contrasts.csv", index=False)

# ============================================================
# 6. 3k vs 12k refinement for k=4, L>=4
# ============================================================

base_k4 = main[(main.k == 4) & (main.L >= 4)].copy()
refine = p1_pub[p1_pub.experiment == "k4_refinement_12k"].copy()

cmp_rows = []
for L in sorted(refine.L.unique()):
    a = base_k4[base_k4.L == L][["seed","model_gap"]]
    b = refine[refine.L == L][["seed","model_gap"]]
    m = a.merge(b, on="seed", suffixes=("_3k","_12k"))

    mean, lo, hi = paired_bootstrap_diff(m.model_gap_3k, m.model_gap_12k)
    cmp_rows.append({
        "L": L,
        "model_gap_3k_mean": m.model_gap_3k.mean(),
        "model_gap_12k_mean": m.model_gap_12k.mean(),
        "mean_reduction": mean,
        "ci95_low": lo,
        "ci95_high": hi,
        "relative_reduction_pct": 100 * (m.model_gap_3k.mean()-m.model_gap_12k.mean()) / m.model_gap_3k.mean(),
        "n": len(m),
    })

refinement_summary = pd.DataFrame(cmp_rows)
refinement_summary.to_csv(CSV_DIR / "part1_k4_refinement_summary.csv", index=False)

# ============================================================
# 7. Robustness tables
# ============================================================

# Capacity
cap = p1_pub[p1_pub.experiment.str.startswith("capacity_")].copy()
cap["capacity"] = cap["experiment"].str.replace("capacity_","", regex=False)

capacity_summary = (
    cap.groupby(["capacity","L"], as_index=False)
       .agg(
           total_excess_mean=("total_excess","mean"),
           info_gap_mean=("info_gap","mean"),
           model_gap_mean=("model_gap","mean"),
           model_gap_sd=("model_gap","std"),
           n=("seed","count")
       )
)
capacity_summary.to_csv(CSV_DIR / "part1_capacity_summary.csv", index=False)

# Sparse
sparse = p1_pub[p1_pub.experiment.isin(["sparse_s2","sparse_s4"])].copy()
sparse["s"] = sparse["experiment"].str.extract(r"sparse_s(\d+)").astype(int)

sparse_summary = (
    sparse.groupby(["s","k","L"], as_index=False)
          .agg(
              total_excess_mean=("total_excess","mean"),
              info_gap_mean=("info_gap","mean"),
              model_gap_mean=("model_gap","mean"),
              model_gap_sd=("model_gap","std"),
              n=("seed","count")
          )
)
sparse_summary.to_csv(CSV_DIR / "part1_sparse_summary.csv", index=False)

# Vocabulary
vocab = p1_pub[p1_pub.experiment.isin(["vocab_V32_k2","vocab_V32_k3"])].copy()
vocab_summary = (
    vocab.groupby(["V","k","L"], as_index=False)
         .agg(
             test_loss_mean=("test_loss","mean"),
             H_star_mean=("H_star","mean"),
             info_gap_mean=("info_gap","mean"),
             model_gap_mean=("model_gap","mean"),
             model_gap_sd=("model_gap","std"),
             n=("seed","count")
         )
)
vocab_summary.to_csv(CSV_DIR / "part1_vocab_summary.csv", index=False)




k4_refinement = refinement_summary
target_zip = resolve_file("V32K3_Targeted_Control_RESULTS.zip")
target = zip_read_csv(target_zip, zip_find_member(target_zip,"v32k3_targeted_scaling_raw.csv"))
if "model_name" not in target:
    target["model_name"] = target.experiment.str.replace("v32k3_target_","",regex=False)
v32_targeted = target.groupby(["model_name","steps"],as_index=False).agg(
    test_loss_mean=("test_loss","mean"),test_loss_sd=("test_loss","std"),
    model_gap_mean=("model_gap","mean"),model_gap_sd=("model_gap","std"),n=("seed","count"))

# Export clean copies into the unified output folder.
tables = {
    "part1_main_threshold_summary.csv": main_summary,
    "part1_threshold_contrasts.csv": threshold_contrasts,
    "part1_k4_refinement_summary.csv": k4_refinement,
    "part1_capacity_summary.csv": capacity_summary,
    "part1_sparse_summary.csv": sparse_summary,
    "part1_vocab_summary.csv": vocab_summary,
    "part1_v32k3_targeted_summary.csv": v32_targeted,
    "part2_final_linear_summary.csv": p2_final_summary,
    "part2_threshold_contrasts.csv": p2_threshold_contrasts,
    "part2_model_size_comparison.csv": p2_model_size,
    "part2_linear_vs_mlp.csv": p2_linear_vs_mlp,
    "part2_input_vs_state_control_summary_final.csv": control_summary,
    "cross_part_synthesis.csv": cross_part,
}
for name, df in tables.items():
    df.to_csv(TABLE_DIR / name, index=False)

print("Loaded and re-exported", len(tables), "paper tables.")
display(main_summary)
display(p2_final_summary.head())
````

## Supplementary true-state and Pythia cross-family figures from archived raw runs


````python

# ============================================================
# 5. Supplementary figures S1–S2 — analysis only
# ============================================================
def bootstrap_mean_ci(values, B=10000, seed=12345):
    values=np.asarray(values,dtype=float)
    values=values[np.isfinite(values)]
    if len(values)==0: return np.nan,np.nan,np.nan
    mean=values.mean()
    if len(values)==1: return mean,mean,mean
    rng=np.random.default_rng(seed)
    samples=rng.choice(values,size=(B,len(values)),replace=True)
    lo,hi=np.quantile(samples.mean(axis=1),[0.025,0.975])
    return mean,lo,hi

def mean_ci_table(df, group_cols, value_col, B=10000):
    rows=[]
    for keys,d in df.groupby(group_cols):
        if not isinstance(keys,tuple): keys=(keys,)
        mean,lo,hi=bootstrap_mean_ci(d[value_col].values,B=B)
        row=dict(zip(group_cols,keys))
        row.update({"mean":mean,"ci_low":lo,"ci_high":hi,"n":d["seed"].nunique()})
        rows.append(row)
    return pd.DataFrame(rows)

# TRUE-STATE raw runs
true_rows=[]
with zipfile.ZipFile(FILES["true_state"],"r") as z:
    for name in z.namelist():
        if name.startswith("runs/") and name.endswith("/true_state_readout.csv"):
            true_rows.append(pd.read_csv(io.BytesIO(z.read(name))))
true_df=pd.concat(true_rows,ignore_index=True)
true_df["k"]=true_df["k"].astype(int); true_df["seed"]=true_df["seed"].astype(int)

# PYTHIA raw runs
pythia_rows=[]
with zipfile.ZipFile(FILES["pythia"],"r") as z:
    for name in z.namelist():
        if name.startswith("runs/") and name.endswith("/probe_results.csv"):
            pythia_rows.append(pd.read_csv(io.BytesIO(z.read(name))))
pythia_df=pd.concat(pythia_rows,ignore_index=True)
pythia_df["k"]=pythia_df["k"].astype(int); pythia_df["L"]=pythia_df["L"].astype(int); pythia_df["seed"]=pythia_df["seed"].astype(int)
pythia_df["relative_context"]=pythia_df["L"]-pythia_df["k"]

s1_summary=mean_ci_table(true_df,["k","probe_type","training_target"],"excess_over_Hstar_test",B=10000)
s2_summary=mean_ci_table(pythia_df,["k","relative_context","probe_type"],"state_exact_acc",B=10000)
s1_summary.to_csv(TABLE_DIR/"supp_true_state_summary_recomputed.csv",index=False)
s2_summary.to_csv(TABLE_DIR/"supp_pythia_summary_recomputed.csv",index=False)

COLORS=plt.rcParams["axes.prop_cycle"].by_key()["color"]
BOUNDARY_COLOR="0.45"
def _style(ax):
    ax.grid(True,axis="y",alpha=0.20,linewidth=0.6)
    ax.spines["top"].set_visible(False); ax.spines["right"].set_visible(False)

# S1
target_order=["hard_sampled_next_token","oracle_transition_distribution"]
probe_order=["linear","mlp"]
target_labels={"hard_sampled_next_token":"Hard next-token targets",
               "oracle_transition_distribution":"Oracle transition targets"}
probe_labels={"linear":"Linear","mlp":"MLP"}
ks=sorted(true_df["k"].unique())
fig,axes=plt.subplots(1,len(ks),figsize=(12.6,3.9),sharey=True)
if len(ks)==1: axes=[axes]
for ax,k in zip(axes,ks):
    dk=s1_summary[s1_summary["k"]==k]; x=np.arange(len(probe_order))
    for i,target in enumerate(target_order):
        d=dk[dk["training_target"]==target].set_index("probe_type").reindex(probe_order)
        y=d["mean"].values; lo=d["ci_low"].values; hi=d["ci_high"].values
        ax.errorbar(x,y,yerr=np.vstack([y-lo,hi-y]),color=COLORS[i],
                    marker="o" if i==0 else "s",linestyle="-" if i==0 else "--",
                    capsize=3,label=target_labels[target],zorder=3)
    ax.axhline(0,color=BOUNDARY_COLOR,linewidth=0.9,zorder=0)
    ax.set_xticks(x); ax.set_xticklabels([probe_labels[p] for p in probe_order])
    ax.set_title(rf"$k={k}$"); ax.set_xlabel("Predictive readout"); _style(ax)
axes[0].set_ylabel(r"Excess CE over $H^*_{\mathrm{test}}$ (nats)")
handles,labels=axes[0].get_legend_handles_labels()
fig.legend(handles,labels,loc="upper center",bbox_to_anchor=(0.5,1.06),ncol=2,frameon=False)
fig.subplots_adjust(top=0.80,bottom=0.18,left=0.09,right=0.99,wspace=0.12)
fig.savefig(FIG_DIR/"FigS1_true_state_predictive_readout_control.png",dpi=300,bbox_inches="tight")
fig.savefig(FIG_DIR/"FigS1_true_state_predictive_readout_control.pdf",bbox_inches="tight")
plt.show()

# S2
probes=["linear","mlp_final"]; probe_labels={"linear":"Linear probe","mlp_final":"MLP probe"}
ks=sorted(pythia_df["k"].unique())
fig,axes=plt.subplots(1,len(ks),figsize=(12.6,3.9),sharex=True,sharey=True)
if len(ks)==1: axes=[axes]
for ax,k in zip(axes,ks):
    for i,probe in enumerate(probes):
        d=s2_summary[(s2_summary["k"]==k)&(s2_summary["probe_type"]==probe)].sort_values("relative_context")
        y=d["mean"].values; lo=d["ci_low"].values; hi=d["ci_high"].values
        ax.errorbar(d["relative_context"],y,yerr=np.vstack([y-lo,hi-y]),color=COLORS[i],
                    marker="o" if i==0 else "s",linestyle="-" if i==0 else "--",
                    capsize=3,label=probe_labels[probe],zorder=3)
    ax.axvline(0,color=BOUNDARY_COLOR,linestyle=":",linewidth=1.25,zorder=0)
    ax.set_title(rf"$k={k}$"); ax.set_xlabel(r"Relative context $L-k$")
    ax.set_xticks([-1,0,1]); ax.set_ylim(-0.02,1.05); _style(ax)
axes[0].set_ylabel("Exact full-state recovery")
handles,labels=axes[0].get_legend_handles_labels()
fig.legend(handles,labels,loc="upper center",bbox_to_anchor=(0.5,1.06),ncol=2,frameon=False)
fig.subplots_adjust(top=0.80,bottom=0.18,left=0.08,right=0.99,wspace=0.10)
fig.savefig(FIG_DIR/"FigS2_pythia_cross_family_validation.png",dpi=300,bbox_inches="tight")
fig.savefig(FIG_DIR/"FigS2_pythia_cross_family_validation.pdf",bbox_inches="tight")
plt.show()

print("SUPPLEMENTARY FIGURES S1–S2: COMPLETE")
````
